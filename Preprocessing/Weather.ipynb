```python
import os
import pandas as pd
import json

# Paths to the folders
flight_base_paths = [r"C:\Users\shivu\ML PROJECT\2016_original_flight", r"C:\Users\shivu\ML PROJECT\2017_original_flight"]
weather_base_path = r"C:\Users\shivu\ML PROJECT\Weather"

# Airport codes to filter
airport_codes = ["ATL", "CLT", "DEN", "DFW", "EWR", "IAH", "JFK", "LAS", "LAX", "MCO", "MIA", "ORD", "PHX", "SEA", "SFO"]

# Recommended columns
flight_columns = [
    "FlightDate", "Quarter", "Year", "Month", "DayofMonth", "DepTime", "DepDel15",
    "CRSDepTime", "DepDelayMinutes", "Origin", "Dest", "ArrTime",
    "CRSArrTime", "ArrDel15", "ArrDelayMinutes"
]
weather_columns = [
    "WindSpeedKmph", "WindDirDegree", "WeatherCode", "precipMM", "Visibilty", "Pressure",
    "Cloudcover", "DewPointF", "WindGustKmph", "tempF", "WindChillF", "Humidity", "date", "time", "airport"
]

# Output paths
weather_csv_path = "D:/weather_data.csv"
```

```python
# Step 1: Process weather data
def process_weather_data():
    weather_data = []
    for airport in airport_codes:
        airport_path = os.path.join(weather_base_path, airport)
        for file_name in os.listdir(airport_path):
            if file_name.startswith("2016") or file_name.startswith("2017"):
                with open(os.path.join(airport_path, file_name), 'r') as file:
                    try:
                        # Load JSON data
                        data = json.load(file)
                        for day_data in data.get("data", {}).get("weather", []):
                            date = day_data.get("date")
                            for hourly_data in day_data.get("hourly", []):
                                record = {
                                    "WindSpeedKmph": hourly_data.get("windspeedKmph"),
                                    "WindDirDegree": hourly_data.get("winddirDegree"),
                                    "WeatherCode": hourly_data.get("weatherCode"),
                                    "precipMM": hourly_data.get("precipMM"),
                                    "Visibilty": hourly_data.get("visibility"),
                                    "Pressure": hourly_data.get("pressure"),
                                    "Cloudcover": hourly_data.get("cloudcover"),
                                    "DewPointF": hourly_data.get("DewPointF"),
                                    "WindGustKmph": hourly_data.get("WindGustKmph"),
                                    "tempF": hourly_data.get("tempF"),
                                    "WindChillF": hourly_data.get("WindChillF"),
                                    "Humidity": hourly_data.get("humidity"),
                                    "date": date,
                                    "time": hourly_data.get("time"),
                                    "airport": airport
                                }
                                weather_data.append(record)
                    except json.JSONDecodeError as e:
                        print(f"Error decoding JSON in file {file_name}: {e}")

    weather_df = pd.DataFrame(weather_data)
    weather_df = weather_df[weather_columns]
    weather_df.to_csv(weather_csv_path, index=False)
    print(f"Weather data saved to {weather_csv_path}")
```

```python
process_weather_data()
```
