```python
import os
import pandas as pd

def combine_flight_data(base_folders, airport_codes, selected_columns):
    """
    Combine CSV files from multiple base folders, filter for specific airport codes
    in the Origin column first, and then filter for the same codes in the Dest column,
    and select specified columns.

    Args:
        base_folders (list): List of paths to base folders containing year folders.
        airport_codes (list): List of airport codes to filter for.
        selected_columns (list): List of columns to retain in the final DataFrame.

    Returns:
        pd.DataFrame: Combined DataFrame with filtered data.
    """
    combined_data = []

    # Iterate through each base folder
    for base_folder in base_folders:
        # Iterate through each year folder
        for year_folder in os.listdir(base_folder):
            year_path = os.path.join(base_folder, year_folder)
            if os.path.isdir(year_path):
                # Iterate through each month's folder
                for month_folder in os.listdir(year_path):
                    month_path = os.path.join(year_path, month_folder)
                    if os.path.isdir(month_path):
                        # Look for CSV files in the folder
                        for file in os.listdir(month_path):
                            if file.endswith('.csv'):
                                file_path = os.path.join(month_path, file)
                                # Read the CSV file
                                df = pd.read_csv(file_path, usecols=selected_columns)
                                # Filter rows based on Origin column first
                                df_filtered_origin = df[df['Origin'].isin(airport_codes)]
                                # Filter rows based on Dest column from the filtered Origin rows
                                df_filtered = df_filtered_origin[df_filtered_origin['Dest'].isin(airport_codes)]
                                combined_data.append(df_filtered)

    # Concatenate all filtered data into a single DataFrame
    return pd.concat(combined_data, ignore_index=True)

# Example usage
selected_columns = [
    "FlightDate", "Quarter", "Year", "Month", "DayofMonth",
    "DepTime", "DepDel15", "CRSDepTime", "DepDelayMinutes",
    "Origin", "Dest", "ArrTime", "CRSArrTime",
    "ArrDel15", "ArrDelayMinutes"
]

airport_codes = ["ATL", "CLT", "DEN", "DFW", "EWR", "IAH", "JFK", "LAS", "LAX", "MCO", "MIA", "ORD", "PHX", "SEA", "SFO"]

# Replace with the paths to your base folders
base_folders = [r"C:\Users\shivu\ML PROJECT\2016_original_flight", r"C:\Users\shivu\ML PROJECT\2017_original_flight"]

# Call the function and get the result
combined_df = combine_flight_data(base_folders, airport_codes, selected_columns)

# Save the result to a CSV file in D drive
combined_df.to_csv("D:\\combined_flight_data.csv", index=False)
```
