```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report,r2_score,mean_squared_error
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import ExtraTreesClassifier,RandomForestClassifier
from sklearn.linear_model import LogisticRegression,LinearRegression
from sklearn.tree import DecisionTreeClassifier
import xgboost as xgb
from sklearn.ensemble import ExtraTreesRegressor,RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python
path = "/content/drive/MyDrive/merged_flight_weather.csv"
```

```python
data = pd.read_csv(path)
```

```python
def time_to_seconds(time_str):
    """Convert time in HHMMSS format to total seconds."""
    try:
        h, m, s = int(time_str[:2]), int(time_str[2:4]), int(time_str[4:])
        return h * 3600 + m * 60 + s
    except ValueError:
        return 0  # Handle invalid time strings

data['CRSDepTime'] = data['CRSDepTime'].astype(str).apply(time_to_seconds)
```

```python
features = ['CRSArrTime', 'WindSpeedKmph', 'WindDirDegree', 'WeatherCode',
            'precipMM', 'Visibilty', 'Pressure', 'Cloudcover', 'DewPointF',
            'WindGustKmph', 'tempF', 'WindChillF', 'Humidity', 'Year',
            'Quarter', 'Month', 'DayofMonth', 'CRSDepTime', 'DepDelayMinutes']
```

```python
data = data.fillna(0)
target_reg = 'ArrDelayMinutes'
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
```

```python
X_reg = data[features]
y_reg = data[target_reg]
```

```python
X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(X_reg, y_reg, test_size=0.3, random_state=42)
```

```python
lin_reg = LinearRegression()
lin_reg.fit(X_reg_train, y_reg_train)
y_pred_lin = lin_reg.predict(X_reg_test)
r2 = r2_score(y_reg_test, y_pred_lin)
mse = mean_squared_error(y_reg_test, y_pred_lin)
rmse = np.sqrt(mse)
print(f"R² Score: {r2:.2f}")
print(f"Root Mean Squared Error: {rmse:.2f}")
```

```python
rfr = RandomForestRegressor(random_state=42, n_estimators=100, max_depth=10)
rfr.fit(X_reg_train, y_reg_train)
y_pred_rfr = rfr.predict(X_reg_test)
r2 = r2_score(y_reg_test, y_pred_rfr)
mse = mean_squared_error(y_reg_test, y_pred_rfr)
rmse = np.sqrt(mse)
print(f"R² Score: {r2:.2f}")
print(f"Mean Squared Error: {mse:.2f}")
print(f"Root Mean Squared Error: {rmse:.2f}")
```

```python
xgr = xgb.XGBRegressor(objective='reg:squarederror', eval_metric='rmse')
xgr.fit(X_reg_train, y_reg_train)
y_pred_xgr = xgr.predict(X_reg_test)
r2 = r2_score(y_reg_test, y_pred_xgr)
mse = mean_squared_error(y_reg_test, y_pred_xgr)
rmse = np.sqrt(mse)
print(f"R² Score: {r2:.2f}")
print(f"Mean Squared Error: {mse:.2f}")
print(f"Root Mean Squared Error: {rmse:.2f}")
```

```python
etr = ExtraTreesRegressor(random_state=42, n_estimators=100, max_depth=10)
etr.fit(X_reg_train, y_reg_train)
y_pred_etc = etr.predict(X_reg_test)
r2 = r2_score(y_reg_test, y_pred_etc)
mse = mean_squared_error(y_reg_test, y_pred_etc)
rmse = np.sqrt(mse)
print(f"R² Score: {r2:.2f}")
print(f"Mean Squared Error: {mse:.2f}")
print(f"Root Mean Squared Error: {rmse:.2f}")
```
